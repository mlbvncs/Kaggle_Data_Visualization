0. Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

1. Carregando os dados

```python
#Caminho do csv lido
iris_filepath = "../data/iris.csv"

#Lendo o arquivo dentro da variável
iris_data = pd.read_csv(iris_filepath, index_col="Id")
```

2. Examinando os dados

```python
#Imprimindo as cinco primeiras linhas do conjunto de dados
iris_data.head()
```

3. Plotando os dados

```python
#Definindo altura e largura da figura
plt.figure(figsize=(16,6))

#Histograma mostrando como a variável petal length varia
#Histograma é usado para visualizar a distribuição de uma variável contínua
sns.histplot(iris_data['Petal Length (cm)'])
```

```python
#Definindo altura e largura da figura
plt.figure(figsize=(16,6))

#Gráfico de densidade mostrando a densidade de pental lenght no eixo x
#Gráfico de densidade é um histograma suavizado
sns.kdeplot(data=iris_data['Petal Length (cm)'], shade=True) #shade colore a área abaixo da linha do gráfico
```

```python
#Definindo altura e largura da figura
plt.figure(figsize=(16,6))

#Gráfico de densidade 2D
'''Gráfico de densidade com a codificação por cores mostrando
a probabilidade de observarmos diferentes combinações de largura 
da sépala e comprimento da pétala, onde as partes mais escuras 
da figura representam maior probabilidade'''
sns.jointplot(x=iris_data['Petal Length (cm)'], y=iris_data['Sepal Width (cm)'], kind="kde")
```

```python
#Definindo altura e largura da figura e título
plt.figure(figsize=(16,6))
plt.title("Histogram of Petal Lengths, by Species")

#Histograma para cada espécie
sns.histplot(data=iris_data, x='Petal Length (cm)', hue='Species')
```

```python
#Definindo altura e largura da figura e título
plt.figure(figsize=(16,6))
plt.title("Distribution of Petal Lengths, by Species")

#Gráfico de densidade para cada espécie, mostrando sua concentração de dados
sns.kdeplot(data=iris_data, x='Petal Length (cm)', hue='Species', shade=True)
```
