0. Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

1. Carregando os dados

```python
#Caminho do csv lido
spotify_filepath = "../data/spotify.csv"

#Lendo o arquivo dentro da variável
spotify_data = pd.read_csv(spotify_filepath, index_col="Date", parse_dates=True)
```

2. Examinando os dados

```python
#Imprimindo as cinco primeiras linhas do conjunto de dados
spotify_data.head()
```

```python
#Imprimindo as últimas cinco linhas do conjunto de dados
#Verificando se tudo está ok, considerando os valores NULL nas primeiras cinco linhas
spotify_data.tail()
```

3. Plotando os dados

```python
#Definindo altura e largura da figura e o título
plt.figure(figsize=(16,6))
plt.title("Daily Global Streams of Popular Songs in 2017-2018")

#Gráfico de linha mostrando como o o número de streams das músicas evoluiu ao longo do tempo
sns.lineplot(data=spotify_data)
```

4. Plotando um subconjunto dos dados

```python
#Carregando o nome das colunas para uma lista
list(spotify_data.columns)
```

```python
#Definindo altura e largura da figura e o título
plt.figure(figsize=(14,6))
plt.title("Daily Global Streams of Popular Songs in 2017-2018")

#Gráfico de linhas mostrando streams globais diários de 'Shape of You'
sns.lineplot(data=spotify_data['Shape of You'], label="Shape of You")

#Gráfico de linhas mostrando streams globais diários de 'Despacito'
sns.lineplot(data=spotify_data['Despacito'], label="Despacito")

#Adicionando rótulo para o eixo horizontal
plt.xlabel("Date")
```
