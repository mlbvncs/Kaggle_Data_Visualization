0. Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

1. Carregando os dados

```python
#Caminho do csv lido
insurance_filepath = "../data/insurance.csv"

#Lendo o arquivo dentro da variável
insurance_data = pd.read_csv(insurance_filepath)
```

2. Examinando os dados

```python
#Imprimindo as cinco primeiras linhas do conjunto de dados
insurance_data.head()
```

3. Plotando os dados

```python
#Definindo altura e largura da figura
plt.figure(figsize=(14,7))

#Gráfico de dispersão é utilizado para mostrar relação entre duas variáveis numéricas ​​contínuas (dados não têm ordem temporal)
#Gráfico de dispersão para ver a relação entre IMC e valor de cobrança
sns.scatterplot(x=insurance_data['bmi'], y=insurance_data['charges'])
```

O gráfico de dispersão acima sugere que o índice de massa corporal (IMC) e os custos de seguro estão positivamente correlacionados, ou seja, clientes com IMC mais alto geralmente tendem a pagar mais em custos de seguro. (Esse padrão faz sentido, já que um IMC alto geralmente está associado a um maior risco de doenças crônicas.)

Para verificar a força dessa relação, é bo adicionar uma linha de regressão, ou seja, a linha que melhor se ajusta aos dados. Fazemos isso alterando o comando para `sns.regplot`.

```python
sns.regplot(x=insurance_data['bmi'], y=insurance_data['charges'])
```

Coeficiente de Correlação (r de Pearson) mede a força e direção da relação linear entre duas variáveis. Varia de -1 a +1:

r = +1 → Correlação positiva perfeita
r = 0 → Sem correlação
r = -1 → Correlação negativa perfeita

```python
#Gráfico de dispersão com três variáveis
#Gráfico mostrando como fumar afeta o relacionamento entre IMC e valor de cobrança
sns.scatterplot(x=insurance_data['bmi'], y=insurance_data['charges'], hue=insurance_data['smoker'])
```

Este gráfico de dispersão mostra que, embora os não fumantes tendam a pagar um pouco mais com o aumento do IMC, os fumantes pagam MUITO mais.

```python
#Adicionando duas linhas de regressão
sns.lmplot(x="bmi", y="charges", hue="smoker", data=insurance_data)
```

```python
#Gráfico de dispersão entre uma variável categórica e uma numérica, para ver a relação entre elas
sns.swarmplot(x=insurance_data['smoker'], y=insurance_data['charges'])
```

Em média, os não fumantes pagam menos do que os fumantes, e os clientes que pagam mais são os fumantes; enquanto os clientes que pagam menos são os não fumantes.
